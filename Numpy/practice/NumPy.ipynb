# What is numpy?

A Python package called NumPy is mostly used for numerical and scientific computations. It offers arrays and several helpful utilities for swiftly carrying out logical and mathematical calculations. Additionally, it is helpful for random number operations, linear algebra, statistics, and data analysis.
ndarray is the primary object in NumPy. Multidimensional arrays, which typically include components of the same data type, are stored in it.

# Numpy Arrays Vs ordered groups of values
Fixed Size: Once a NumPy array is created, its size remains fixed. If you need to change the size, a new array has to be created.

Same Data Type: The elements in a NumPy array generally have the same data type and occupy the same amount of memory.

Fast Operations: NumPy allows mathematical operations on large amounts of data to be performed quickly with less code.

Widely Used: NumPy arrays are commonly used for data processing in many scientific and data-related Python libraries.

### np.ceil() → Ceiling

Definition: Rounds each number UP to the nearest integer.

Always goes toward +∞, no matter what.

```python
import numpy as np

arr = np.array([1.2, 2.7, -1.5, -2.3])
arr
```

```python
np.ceil(arr)
```

### np.floor() → Floor

Definition: Rounds each number DOWN to the nearest integer.

Always goes toward -∞.

```python
np.floor(arr)
```

### np.round() → Round (to nearest)

Definition: Rounds each number to the nearest integer.

If decimal = 0.5 → rounds to nearest even integer (NumPy behavior).

```python
np.round(arr)
```
